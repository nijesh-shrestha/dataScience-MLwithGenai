```python
import numpy as np

A = np.array([[1, 2], [3, 4]])
det_A = np.linalg.det(A)
print(det_A)
```

```python
#2x + y = 5,
#x -y = 1

a = np.array([[2, 1], [1, -1]])
b = np.array([5, 1])

solutions = np.linalg.solve(a, b)
print("Solutions = ", solutions)
```

```python
#3x + y -2z = 9
#2x - y + z = 3
#x +2y + 3z = 12

x = np.array([[3, 1, -2], [2, -1, 1], [1, 2, 3]])
y = np.array([9, 3, 12])

ans = np.linalg.solve(x, y)
print("Ans = ", ans)
```
