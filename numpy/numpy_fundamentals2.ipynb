```python
import numpy as np
```

```python
x = np.array([[1, 2], [3, 4]], dtype=np.float64)
y = np.array([[5, 6], [7, 8]], dtype=np.float64)

print(x + y)
print(np.add(x, y))
```

```python
print(np.dot(x, y))
```

```python
preds = np.array([0.2, 0.5, 0.92, 0.9, 0.1])

to_predict_class = ["apple", "banana", "ball", "mango", "orange"]
possible_class_index = np.where(preds > 0.85)[0]
print(possible_class_index)
```

```python
arr = np.array([1, 4, 7, 2, 8, 4, 9, 10, 10, 8])

print(np.argmax(arr))
print(np.argmin(arr))
```

```python
print(np.where(arr >= arr[np.argmax(arr)])[0])
```

```python
print(np.max(arr))
print(np.min(arr))
print(np.sum(arr))
```

```python
xyz = np.array([[1, 2, 8], [5, 3, 1], [4, 6, 5]])

print(xyz)

print(np.max(xyz, axis=0)) #compute max of each column
print(np.max(xyz, axis=1)) #compute max of each row
```
