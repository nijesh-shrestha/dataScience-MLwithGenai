```python
df.drop(df.columns[0], axis=1, inplace=True)
df
```

```python
df.TV.corr(df.sales)
```

```python
from sklearn.linear_model import LinearRegression

feature_cols = ['TV', 'radio', 'newspaper']
X = df[feature_cols]
y = df.sales

lml = LinearRegression()
lml.fit(X, y)

print(lml.intercept_)
print(lml.coef_)
```

```python
list(zip(feature_cols, lml.coef_))
```

```python
lml_preds = lml.predict(X[['TV', 'radio', 'newspaper']])
```

```python
from sklearn.metrics import mean_squared_error, r2_score

print("R2S :", r2_score(y, lml_preds))
print("MSE :", mean_squared_error(y, lml_preds))
```
