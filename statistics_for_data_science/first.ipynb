```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
```

```python
data = np.random.normal(50, 10, 1000)

print(f"Mean: {np.mean(data):.2f}")
print(f"Median: {np.median(data):.2f}")
print(f"Standard Deviation: {np.std(data):.2f}")
print(f"Minimum: {np.min(data):.2f}")
print(f"Maximum: {np.max(data):.2f}")

plt.hist(data, bins=30, edgecolor='black')
plt.title('Data Distribution')
plt.xlabel('Values')
plt.ylabel('Frequency')
plt.show()
```

```python
data2 = [2, 4, 6, 8]
var_pop = np.var(data2, ddof=0)
print("Population Variance:", var_pop)

var_sample = np.var(data, ddof=1)
print("Sample Variance:", var_sample)
```

```python
np.random.seed(42)
stock_A = np.random.normal(1.5, 0.5, 100)
stock_B = np.random.normal(1.5, 2.0, 100)

plt.figure(figsize=(10, 5))
plt.plot(stock_A, label='Stock A (SD=0.5)', alpha=0.7)
plt.plot(stock_B, label='Stock B (SD=2.0)', alpha=0.7)
plt.axhline(1.5, color='red', linestyle='--', label='Mean (1.5%)')
plt.title("Stock Volatility Comparison")
plt.ylabel('Daily Return (%)')
plt.legend()
plt.show()
```
